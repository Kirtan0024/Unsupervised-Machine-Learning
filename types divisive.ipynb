## Task 1

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import linkage, dendrogram

iris = load_iris()
X = iris.data

Z_ward = linkage(X, method="ward")

plt.figure(figsize=(12, 6))
dendrogram(Z_ward)
plt.title("Iris Dataset - Ward Linkage")
plt.xlabel("Samples")
plt.ylabel("Distance")
plt.show()
```

## Task 2

```python
from scipy.cluster.hierarchy import fcluster

cluster_height = 100
clusters = fcluster(Z_ward, t=cluster_height, criterion="distance")

cluster_counts = pd.Series(clusters).value_counts().sort_index()

print("Cluster assignments:")
print(clusters)

print("\nNumber of samples in each cluster:")
print(cluster_counts)
```

## Task 3

```python
Z_single = linkage(X, method="single")
Z_complete = linkage(X, method="complete")

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

dendrogram(Z_ward, ax=axes[0], no_labels=True)
axes[0].set_title("Ward Linkage")
axes[0].set_xlabel("Samples")
axes[0].set_ylabel("Distance")

dendrogram(Z_single, ax=axes[1], no_labels=True)
axes[1].set_title("Single Linkage")
axes[1].set_xlabel("Samples")
axes[1].set_ylabel("Distance")

dendrogram(Z_complete, ax=axes[2], no_labels=True)
axes[2].set_title("Complete Linkage")
axes[2].set_xlabel("Samples")
axes[2].set_ylabel("Distance")

plt.tight_layout()
plt.show()
```

Ward linkage generally creates compact and balanced groups by m

## Task 4

Clusters that merge at a lower height are more similar because only a small distance separates them. Clusters that remain separate until a higher height are more different from each other. Large vertical gaps before a merge can indicate natural boundaries between groups. The overall branch structure shows how individual samples gradually combine into larger clusters.
